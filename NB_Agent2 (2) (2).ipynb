```python

#@title Start daemon on `default` (GPU if available)
import os, sys, subprocess, shlex, time
cmd = "clearml-agent daemon --queue controller --create-queue "
print("Starting agent:", cmd)
process = subprocess.Popen(shlex.split(cmd), stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
try:
    for _ in range(120):
        line = process.stdout.readline()
        if not line:
            time.sleep(1)
            continue
        print(line.rstrip())
except KeyboardInterrupt:
    process.terminate()
    print("Agent terminated.")
```
